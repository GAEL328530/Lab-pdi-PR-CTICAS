Ejercicio 1

Primero importo las librerías que voy a usar: OpenCV para trabajar con la imagen, NumPy para la matriz y Matplotlib para mostrar los resultados.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

Cargo la imagen en escala de grises y saco sus dimensiones (alto y ancho). Si no la encuentra, el programa avisa para no seguir con errores confusos.

```python
imagen = cv2.imread("/content/cerebrito.jpg", cv2.IMREAD_GRAYSCALE)

if imagen is None:
    raise FileNotFoundError("No se encontró /content/cerebrito.jpg, hay que subirla primero")

alto, ancho = imagen.shape
```

Defino la matriz de traslación para mover la imagen 50 px a la derecha y 30 px hacia abajo, y la aplico con una transformación afín manteniendo el tamaño original.

```python
tx, ty = 50, 30
M = np.float32([[1, 0, tx], [0, 1, ty]])

imagen_trasladada = cv2.warpAffine(imagen, M, (ancho, alto))
```

Por último, muestro la imagen original y la trasladada lado a lado para comparar el desplazamiento.

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(imagen, cmap="gray")
plt.title("Imagen Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(imagen_trasladada, cmap="gray")
plt.title("Imagen Trasladada")
plt.axis("off")

plt.show()
```

Defino la matriz de traslación con valores decimales (100.5 px en X y 100.7 px en Y). Como el desplazamiento no cae en píxeles exactos, uso interpolación lineal (INTER_LINEAR) para calcular los valores intermedios al aplicar la transformación.

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(imagen, cmap="gray")
plt.title("Imagen Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(imagen_trasladada, cmap="gray")
plt.title("Imagen Trasladada (Decimales)")
plt.axis("off")

plt.show()
```

Importo las librerías que voy a usar: OpenCV para trabajar con la imagen, NumPy para las matrices y Matplotlib para mostrar los resultados.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

Cargo la imagen en escala de grises y saco sus dimensiones (alto y ancho). Si no la encuentra, el programa avisa para no seguir con errores confusos.

```python
imagen = cv2.imread("/content/cerebrito.jpg", cv2.IMREAD_GRAYSCALE)

if imagen is None:
    raise FileNotFoundError("No se encontró /content/cerebrito.jpg, hay que subirla primero")

alto, ancho = imagen.shape
```

Aplico la rotación de 45° alrededor del centro de la imagen, sin cambiar la escala. Primero calculo el centro, luego obtengo la matriz de rotación y por último la aplico manteniendo el tamaño original.

```python
angulo = 45
centro = (ancho // 2, alto // 2)
escala = 1.0

M_rotacion = cv2.getRotationMatrix2D(centro, angulo, escala)

imagen_rotada = cv2.warpAffine(imagen, M_rotacion, (ancho, alto))
```

Por último, muestro la imagen original y la rotada lado a lado para comparar el giro.

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(imagen, cmap="gray")
plt.title("Imagen Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(imagen_rotada, cmap="gray")
plt.title("Imagen Rotada 45°")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
```

Cargo la imagen en escala de grises. Si no la encuentra, el programa avisa para no seguir con errores confusos.

```python
imagen = cv2.imread("/content/cerebrito.jpg", cv2.IMREAD_GRAYSCALE)

if imagen is None:
    raise FileNotFoundError("No se encontró /content/cerebrito.jpg, hay que subirla primero")
```

Aplico los dos escalados con la función resize: uno al 150 % para agrandar la imagen y otro al 50 % para reducirla. Para agrandar uso interpolación lineal, y para reducir uso INTER_AREA, que da mejores resultados al encoger.

```python
imagen_150 = cv2.resize(imagen, None, fx=1.5, fy=1.5, interpolation=cv2.INTER_LINEAR)

imagen_50 = cv2.resize(imagen, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_AREA)
```

Por último, muestro la imagen original y las dos escaladas lado a lado. Dejo los ejes visibles y pongo las dimensiones en cada título para que se note la diferencia de tamaño entre ellas.

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(imagen, cmap="gray")
plt.title(f"Imagen Original {imagen.shape[1]}x{imagen.shape[0]}")

plt.subplot(1, 3, 2)
plt.imshow(imagen_150, cmap="gray")
plt.title(f"Escala 150% {imagen_150.shape[1]}x{imagen_150.shape[0]}")

plt.subplot(1, 3, 3)
plt.imshow(imagen_50, cmap="gray")
plt.title(f"Escala 50% {imagen_50.shape[1]}x{imagen_50.shape[0]}")

plt.show()
```

Importo las librerías que voy a usar: OpenCV para trabajar con la imagen y Matplotlib para mostrar las imágenes y los histogramas.

```python
import cv2
import matplotlib.pyplot as plt
```

Cargo la imagen médica en escala de grises. Si no la encuentra, el programa avisa para no seguir con errores confusos.

```python
imagen = cv2.imread("/content/cerebrito.jpg", cv2.IMREAD_GRAYSCALE)

if imagen is None:
    raise FileNotFoundError("No se encontró /content/cerebrito.jpg, hay que subirla primero")
```

Obtengo el histograma de la imagen original, que muestra cuántos píxeles hay en cada nivel de gris (de 0 a 255). En una imagen de bajo contraste, los valores suelen quedar amontonados en una zona estrecha.

```python
hist_original = cv2.calcHist([imagen], [0], None, [256], [0, 256])
```

Aplico la ecualización de histograma, que redistribuye los niveles de gris para aprovechar todo el rango disponible y así mejorar el contraste. Después calculo el histograma de la imagen mejorada.

```python
imagen_ecualizada = cv2.equalizeHist(imagen)

hist_ecualizado = cv2.calcHist([imagen_ecualizada], [0], None, [256], [0, 256])
```

Por último, muestro en la fila de arriba la imagen original y la mejorada, y en la fila de abajo sus histogramas, para comparar cómo cambió la distribución de los grises.

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(imagen, cmap="gray", vmin=0, vmax=255)
plt.title("Imagen Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(imagen_ecualizada, cmap="gray", vmin=0, vmax=255)
plt.title("Imagen Ecualizada")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.plot(hist_original, color="black")
plt.title("Histograma Original")
plt.xlabel("Nivel de gris")
plt.ylabel("Número de píxeles")
plt.xlim([0, 255])

plt.subplot(2, 2, 4)
plt.plot(hist_ecualizado, color="black")
plt.title("Histograma Ecualizado")
plt.xlabel("Nivel de gris")
plt.ylabel("Número de píxeles")
plt.xlim([0, 255])

plt.tight_layout()
plt.show()
```
